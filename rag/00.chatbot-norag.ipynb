## Building a Chatbot

### Import Libraries

```python
from langchain_ollama import ChatOllama
from IPython.display import Markdown
```

```python
from langchain_ollama import ChatOllama

model = ChatOllama(
    model="llama3.1:8b",
    temperature=0
)
```

### Create an LLM Call

This function defines a system prompt to guide the LLM. It takes a user prompt as input, calls the model, and returns the response.

```python
def llm_call(user_prompt: str):

    prompt = [
        {
            "role": "system",
            "content": """
You are an AI Medicine Reminder and Health Assistant.

Your responsibilities are:
- Answer medicine-related questions clearly.
- Explain medicine usage, dosage, precautions, and possible side effects.
- Provide general health guidance.
- Remind users that serious symptoms require consultation with a qualified healthcare professional.
- Do not prescribe medicines or provide a medical diagnosis.
- Keep answers accurate, simple, and easy to understand.
"""
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    return model.invoke(prompt)
```

Let's start querying the LLM

```python
response = llm_call("What are the side effects of Paracetamol?")

print(response.content)
```

Now let's ask questions that may be outside the model's knowledge base. You may notice that the model can produce incorrect or made-up answers (hallucinations).

```python
response = llm_call("What are the symptoms of an overdose of Paracetamol?")
print(response.content)
```

```python
response1 = llm_call("Remind me to take my Paracetamol tablet every day at 8 AM.")
print(response1.content)
```

```python
response8 = llm_call("What medicines am I supposed to take today?")
print(response8.content)
```

```python
response6 = llm_call("Remind me to take my blood pressure tablet after breakfast every day.")
print(response6.content)
```

```python
response = llm_call("What medicines should not be taken during pregnancy?")
print(response.content)
```

```python
response = llm_call("How should Insulin be stored?")
print(response.content)
```
