### # 🎙️ Voice-Based Medicine Reminder System

````python
# ==========================================================
# AI Medicine Reminder - Voice Reminder System
# ==========================================================

# Install (run once)
# !pip install SpeechRecognition pyttsx3 schedule pyaudio

import speech_recognition as sr
import pyttsx3
import schedule
import time
import re

# ----------------------------------------------------------
# Text-to-Speech Engine
# ----------------------------------------------------------

engine = pyttsx3.init()

def speak(text):
    print("AI:", text)
    engine.say(text)
    engine.runAndWait()


# ----------------------------------------------------------
# Speech Recognition
# ----------------------------------------------------------

recognizer = sr.Recognizer()

def listen():
    with sr.Microphone() as source:
        recognizer.adjust_for_ambient_noise(source, duration=1)

        print("🎤 Speak now...")
        audio = recognizer.listen(source)

    try:
        text = recognizer.recognize_google(audio)

        print("You:", text)

        return text.lower()

    except Exception:
        speak("Sorry, I couldn't understand your voice.")
        return ""


# ----------------------------------------------------------
# Reminder Storage
# ----------------------------------------------------------

reminders = []


# ----------------------------------------------------------
# Add Reminder
# ----------------------------------------------------------

def add_reminder(medicine, reminder_time):

    reminders.append({
        "medicine": medicine,
        "time": reminder_time
    })

    schedule.every().day.at(reminder_time).do(
        reminder_notification,
        medicine
    )

    speak(f"Reminder set for {medicine} at {reminder_time} every day.")


# ----------------------------------------------------------
# Reminder Notification
# ----------------------------------------------------------

def reminder_notification(medicine):

    message = f"""
    Reminder!

    It is time to take your {medicine}.

    Please take your medicine after food if prescribed.

    Drink plenty of water.

    Stay healthy.
    """

    speak(message)


# ----------------------------------------------------------
# Extract Medicine and Time
# Example:
# "Remind me to take paracetamol at 09:00"
# ----------------------------------------------------------

def process_command(command):

    medicine_match = re.search(r"take (.*?) at", command)

    time_match = re.search(r"at (\d{2}:\d{2})", command)

    if medicine_match and time_match:

        medicine = medicine_match.group(1)

        reminder_time = time_match.group(1)

        add_reminder(medicine, reminder_time)

    else:

        speak("Please say the medicine and time clearly.")


# ----------------------------------------------------------
# Main Program
# ----------------------------------------------------------

speak("Welcome to AI Medicine Reminder.")

voice = listen()

process_command(voice)

print("\nMedicine Reminder Service Started...\n")

while True:

    schedule.run_pending()

    time.sleep(1)
````
